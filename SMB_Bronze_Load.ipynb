```python
# load raw csv
df = spark.read.option("header", True).csv("Files/Bronze/smb_accounts_bronze.csv")
display(df)
```

```python
df.printSchema()
```

```python
df.count()
```

```python
df.write.mode("overwrite").saveAsTable("bronze_smb_accounts")
```

```python
spark.sql("SELECT COUNT(*) AS row_count FROM bronze_smb_accounts").show()
display(spark.sql("SELECT * FROM bronze_smb_accounts LIMIT 20"))
```

```python
display(spark.sql("""
SELECT DISTINCT industry
FROM bronze_smb_accounts
ORDER BY industry
"""))

display(spark.sql("""
SELECT DISTINCT revenue_tier
FROM bronze_smb_accounts
ORDER BY revenue_tier
"""))
```

```python
display(spark.sql("""
SELECT *
FROM bronze_smb_accounts
WHERE support_tickets = 99
LIMIT 20
"""))

display(spark.sql("""
SELECT *
FROM bronze_smb_accounts
WHERE azure_compute_hours < 0
LIMIT 20
"""))

display(spark.sql("""
SELECT *
FROM bronze_smb_accounts
WHERE m365_licensed_seats < 0 OR m365_licensed_seats > 1000
LIMIT 20
"""))
```

```python
display(spark.sql("""
SELECT DISTINCT month
FROM bronze_smb_accounts
ORDER BY month
"""))
```

```python
display(spark.sql("""
SELECT
    SUM(CASE WHEN account_id IS NULL THEN 1 ELSE 0 END) AS null_account_id,
    SUM(CASE WHEN m365_active_users IS NULL THEN 1 ELSE 0 END) AS null_active_users,
    SUM(CASE WHEN azure_compute_hours IS NULL THEN 1 ELSE 0 END) AS null_azure_hours
FROM bronze_smb_accounts
"""))
```

```python
display(spark.sql("""
SELECT
    MIN(employee_count) AS min_employee_count,
    MAX(employee_count) AS max_employee_count,
    MIN(estimated_clv_12mo) AS min_clv,
    MAX(estimated_clv_12mo) AS max_clv
FROM bronze_smb_accounts
"""))
```
